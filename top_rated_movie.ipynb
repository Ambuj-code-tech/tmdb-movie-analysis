<h1>Top Rated Movie</h1>
<span style="font-size: 0.98rem; color: gray">Using API calling from <a href="https://developer.themoviedb.org/reference/guest-session-rated-movies"> TMDB </a></span>

```python
import os
import pandas as pd
import requests
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
API_KEY = os.getenv("TMDB_API_KEY")
```

```python
data_list = []
```

```python
url = "https://api.themoviedb.org/3/movie/top_rated"

params = {
    "api_key": API_KEY,
    "language": "en-US",
    "page": 1,
}

response = requests.get(url, params=params, timeout=15)
response.raise_for_status()
```

```python
data = response.json()
total_pages = min(data["total_pages"],500)

for page in range(1, total_pages+1):
    params["page"] = page

    response = requests.get(url, params=params, timeout=15)
    response.raise_for_status()

    data_list.extend(response.json()["results"])

df = pd.DataFrame(data_list)

df.head()
```

```python
df.columns
```

```python
columns = ["id", "title", "overview", "release_date", "popularity", "vote_average", "vote_count"]
df = df[columns]

df.head()
```

```python
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]} \n")

print(f"Check for null values\n{df.isnull().sum()}\n")
print(f"Check for duplicated values\n{df.duplicated().sum()}")
```
